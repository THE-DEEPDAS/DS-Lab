```python
import pandas as pd
df = pd.read_csv('Iris.csv')

mean_sepal_setosa = df[df['Species'] == 'Iris-setosa']['SepalLengthCm'].mean()
print(mean_sepal_setosa)
```

```python
import numpy as np
variance_sepal_setosa = df[df['Species'] == 'Iris-setosa']['SepalLengthCm'].var()
sd_sepal_setosa = np.sqrt(variance_sepal_setosa)
print(sd_sepal_setosa)
```

```python
mean_petal_versicolor = df[df['Species'] == 'Iris-versicolor']['PetalLengthCm'].mean()
mean_petal_virginica = df[df['Species'] == 'Iris-virginica']['PetalLengthCm'].mean()
sd_petal_versicolor = np.sqrt(df[df['Species'] == 'Iris-versicolor']['PetalLengthCm'].var())
sd_petal_virginica = np.sqrt(df[df['Species'] == 'Iris-virginica']['PetalLengthCm'].var())
```

```python
print(mean_petal_versicolor, sd_petal_versicolor)
print(mean_petal_virginica, sd_petal_virginica)
```
